save model ->save the model's parameter

```python
class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.hidden=nn.Linear(20,256)
        self.output=nn.Linear(256,10)
    def forward(self,x):
        return F.relu(self.output(self.hidden(x)))

net=MLP()
x=torch.rand(2,20)
y=net(x)
```

```python
#save model's state_dict() (parameters as well)
torch.save(net.state_dict(),'mlp.params')

#load the model
clone=MLP()
clone.load_state_dict(torch.load('mlp.params')) #write back the saved parameters to new model "clone"
                                                #use function "load_state_dict()"
clone.eval()  #change to evalaute mode (for reasoning mode)



#RESONING MODE(TEST MODE):
# with torch.no_grad():
#    output=clone(test_data)
```
