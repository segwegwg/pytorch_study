```python
import torch
from torch import nn
from torch.nn import functional as F
```

```python
def comp_conv2d(conv2d,x):
   
    #nn.Conv2d's parameters must be a 4d tuple
    #so reshape 2d tensor x to 4d
    #tuple can use + to combine 、* to calculate
    
    # do not use x.shape(),because the shape of a tensor is not callable
    #itis attribute rather than a method(because the object is a tensor)
    x=x.reshape((1,1)+x.shape)  #2d24d  
    y=conv2d(x)

    # nn.Conv2d's output---y is a  4d tuple  (batch_size, channels, height, width)
    # we only need height and width(2,3)
    #so do reshape to restore y to 2d
    return y.reshape(y.shape[2:])  #4d22d

#the parameters of nn.Conv2d is(in_channels,out_channels,ks,padding)
#the parameters of reshape is(batch_size,channels,row,column)
conv2d=nn.Conv2d(in_channels=1,out_channels=1,kernel_size=3,padding=1)
x=torch.rand(size=(8,8))
comp_conv2d(conv2d,x).shape



```

## nn.Conv2d 的迷惑行为：  
### 1. 定义的时候的参数： 
输入输出通道数 in_channels/out_channels  
卷积核kernel_size  
填充类型padding  
是否偏置bias=False   
etc.
### 2.对于输入的要求：
必须是一个4d的tuple,包含：  
batch_size  
channels  
输入图片矩阵hight
输入图片矩阵的width

### 由于基本的tensor矩阵只有hight&width两维，所以需要reshape在4d-2d之间相互转换
 

```python
# Asymmetric：
conv2d=nn.Conv2d(in_channels=1,out_channels=1,kernel_size=(5,3),padding=1,stride=(2,3))
comp_conv2d(conv2d,x).shape
```
