### 线性回归（pytorch版本）
#### 固定模板：
* 准备数据
* 定义训练模型类
* 定义损失函数与优化器
* 进行训练：    
                      </br>前向传播：预测-计算误差  
                      </br>反向传播：梯度清零-计算梯度（损失函数偏导数）-参数更新

```python
import torch
```

```python
#导入数据
x=torch.tensor([[1.0],[2.0],[3.0]])
y=torch.tensor([[2.0],[4.0],[6.0]])
```

```python
#定义模型类
class LinearModel(torch.nn.Module):#继承torch库中的神经网络库nn中的Module类
    def __init__(self):#注意双下划线
        super(LinearModel,self).__init__()#注意双下划线
        self.linear=torch.nn.Linear(1,1)
        #创建一个输入特征维度和输出特征维度
        #前一层的输出维度与下一层的输入维度需要一致
        #此行实现线性变换层的构建（.nn.Linear()），输入输出维度一旦确定则权重矩阵与bias矩阵的形状也确定了
        #θ b 都在此行确定   in x out
    def forward(self,x):
        y_pred=self.linear(x) #执行线性变换 y_pred=x.T@ w+b
        return y_pred
        
        
        
```

#### 注意点
* python在oop的类定义中对于函数成员中使用数据成员的时候需要使用self代表实例，而cpp中通常直接操作数据成员（隐藏this指针）
* super函数是借助MRO机制（防止钻石路径调用出错），指的是self这个实例的MRO链中从LinearModel类之后第一个实现_init_的父类,然后再调用那个父类的init函数
* LinearModel类就是正在定义的类，而MRO链中的父类就是继承的torch.nn.Module类，而不直接调用却用super调用就是防止出现钻石路径调用出错


```python
#实例化模型
model=LinearModel()#模型类实例化
criterion=torch.nn.MSELoss(size_average=False)
#定义损失函数为均方误差不平均，也就是只求和

optimizer=torch.optim.SGD(model.parameters(),lr=0.01)
#定义优化器：基于损失梯度更新模型参数
#model.parameters()传入模型对象的所有可训练参数θ（w）,b，控制学习率
#SGD为小批量梯度下降（因为需要对θ中每个维度（wi）进行梯度下降）



for epoch in range(1000):
    y_pred=model(x)
    loss=criterion(y_pred,y)
    print(epoch,loss.item())

    #梯度更新三部曲：
    optimizer.zero_grad()#由于pytorch梯度累计，所以需要先清除for循环上一轮的梯度
    loss.backward()#计算当前loss函数对w与b的偏导数（会代入当前的数值）
    optimizer.step()#更新参数w=w-lr x w_grad

print("w=",model.linear.weight.item())
print("b=",model.linear.bias.item())

```

```python
#预测一下：
x_test=torch.tensor([[4.0]])
y_test=model(x_test)
print("y_pred=",y_test)
```
