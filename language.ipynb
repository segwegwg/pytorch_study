#### 第一步，导入张量tensor

```python
#1.直接导入
a=[[1,2],[3,4],[5,6]]
a_tensor=torch.tensor(a)
```

```python
#2.从numpy转来，本质共享内存位置，同步改变
b=np.array([[1,2],[3,4],[5,6]])
b_tensor=torch.from_numpy(b)
```

```python
a_tensor,b_tensor
```

```python
#3.仿照或者重写自一个已存在的tensor
c_tensor=torch.ones_like(a_tensor)  #构建一个和a_tensor同形状的tensor
c_tensor
```

```python
d_tensor=torch.rand_like(b_tensor,dtype=float)    #构建一个和b_tensor同形状的数据类型为浮点数的张量
d_tensor
```

#### 第三，tensor操作

```python
tensor=torch.rand(3,3)
```

##### 1.装载

```python
#1.将内存中的张量装载到gpu上
#识别当前加速器类型
device=torch.accelerator.current_accelerator().type if torch.accelerator.is_available() else 'cpu'
tensor=tensor.to(device)
print(f"Device tensor is stored on:{tensor.device}")
```

##### 2.设置张量

```python
#2.设置张量的行列值
h=torch.ones(3,3)

i=h
r=torch.clone(h)

h[:,1]=114
i[2,2]=514
i,h,r
```

#### 这里i与h内容完全一致，是因为pytorch的=符号相当于i、h共享同一个张量内存
##### 而如果想克隆（复制）同样的张量，则需要使用.clone()

##### 3.张量拼接

```python
#3.张量拼接
#dim=1横向拼接
#dim=0纵向拼接
r=torch.cat([h,h,h],dim=1)
s=torch.cat([h,h,h],dim=0)#即使是纵向拼接也给出张量列表即可，列表只是给出拼接的张量名字，而非最终的拼接形式
r,s
```

##### 4.张量相乘（基本与numpy一致）
* *同形状张量逐元素相乘,相乘结果仍为该元素（cnn rnn常用）
* @ matmul矩阵乘法

```python
print(a)
print(f"a_tensor*a_tensor:{a_tensor*a_tensor}")
print(f"a_tensor@a_tensor:{a_tensor@a_tensor.T}")##一定注意转置
print(f"matmul:{torch.matmul(a_tensor,a_tensor.T)}")

#matmul另一种调用形式
print(f"matmul second:{a_tensor.matmul(a_tensor.T)}")
```
